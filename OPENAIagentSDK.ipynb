```python
from agents import enable_verbose_stdout_logging,function_tool
enable_verbose_stdout_logging()
@function_tool
def get_current_weather(city:str)-> str:
  """Return weather for the city

  Args:
    city: City name

  Returns:
    Weather

  """
  return f"{city} weather is sunny"
agent:Agent=Agent(name='Assistant',instructions='You are an helpful assistant',model="gemini-2.0-flash",tools=[get_current_weather]
                  )
result=Runner.run_sync(agent,"what is the weather in karachi")
print(result.final_output)
```

```python
from agents import enable_verbose_stdout_logging,function_tool
enable_verbose_stdout_logging()
@function_tool
def get_current_news(industry:str)-> str:
  """Return current news for a particular industry

  Args:
    industry: The industry to look in current news

  Returns:
    news

  """
  return f"{industry } have new concept Agent Native Cloud"
agent:Agent=Agent(name='Assistant',instructions='You are an helpful assistant',model="gemini-2.0-flash",tools=[get_current_weather,get_current_news]
                  )
result=Runner.run_sync(agent,"what is the weather in karachi and share the latest news in tech")
print(result.final_output)
```
